```python
%%sql
select
p.categoryname,
sum(quantity*netprice*exchangerate) as total_rev,
sum(case when s.orderdate between '2022-01-01' and '2022-12-31'then s.productkey end) as Year_sales_2022,
sum(case when s.orderdate between '2023-01-01' and '2023-12-31' then  s.productkey end) as Year_sales_2023

from sales s
left join product p on p.productkey=s.productkey
group by p.categoryname
order by p.categoryname
```

```python
%%sql
select
p.categoryname,
sum(quantity*netprice*exchangerate) as total_rev,
sum(case when s.orderdate between '2022-01-01' and '2022-12-31'then  sum(quantity*netprice*exchangerate) end) as Year_sales_2022,
sum(case when s.orderdate between '2023-01-01' and '2023-12-31' then  s.productkey end) as Year_sales_2023

from sales s
left join product p on p.productkey=s.productkey
group by p.categoryname
order by p.categoryname
```

```python
%%sql
select
p.categoryname,
sum(quantity*netprice*exchangerate) as total_rev,
sum(case when s.orderdate between '2022-01-01' and '2022-12-31' then  (quantity*netprice*exchangerate) end) as Year_sales_2022,
sum(case when s.orderdate between '2023-01-01' and '2023-12-31' then (quantity*netprice*exchangerate) end) as Year_sales_2023

from sales s
left join product p on p.productkey=s.productkey
group by p.categoryname
order by p.categoryname
```
